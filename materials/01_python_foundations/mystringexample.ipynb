checking if a string is palindrome

```python
def palindrome(s):
    return(s==reverse(s))
```

```python
palindrome("malayalam")
```

```python
palindrome("Nitesh")
```

```python
palindrome("odddo")
```

Removing all the occurence from the set of characters from a string

```python
def excludeset(s,pset):
    output=""
    for c in s:
        if not (c in pset):
            output=output+c
    return(output)
```

```python
w="12345678vcdsdfgfhbvdcsa"
```

```python
excludeset(w,"1234567")
```
