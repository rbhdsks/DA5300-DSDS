# Python Arrays

This notebook introduces:
1. Low-Level Arrays (`array.array`)
2. Referential Arrays (`list`)
3. Fixed-size Arrays using `ctypes.py_object`

All examples use a cricket theme.

## 1. Low-Level Arrays
A low-level array stores elements of a single fixed data type.

````
from array import array

jersey = array('i', [18,45,63,1,33,77,8,99,90,93,19])
print(jersey)
````

````
print("First Player:", jersey[0])
print("Last Player:", jersey[-1])
print("Total Players:", len(jersey))
````

````
jersey[3]=17
print(jersey)
````

````
print("Average Jersey:", sum(jersey)/len(jersey))
````

## 2. Referential Arrays
Python lists store **references** to objects.

````
class Player:
    def __init__(self,name,jersey):
        self.name=name
        self.jersey=jersey
    def display(self):
        print(self.name,self.jersey)

virat=Player("Virat",18)
rohit=Player("Rohit",45)
bumrah=Player("Bumrah",93)
````

````
players=[virat,rohit,bumrah]
for p in players:
    p.display()
````

````
print("id(virat):",id(virat))
print("id(players[0]):",id(players[0]))
print("id(rohit):",id(rohit))
print("id(players[1]):",id(players[1]))
````

````
virat.name="Virat Kohli"
players[0].display()
````

````
print("List id:",id(players))
for i,p in enumerate(players):
    print(i,id(p))
````

Notice that `virat` and `players[0]` refer to the same object.

## 3. Fixed-Size Arrays using ctypes
Python lists resize automatically. Here we manually allocate an array of references.

````
import ctypes

capacity=5
A=(ctypes.py_object*capacity)()
print(A)
````

````
A[0]=virat
A[1]=rohit
A[2]=bumrah

for i in range(3):
    print(A[i].name)
````

````
print(id(A[0]))
print(id(virat))
````

````
virat.name="King Kohli"
print(A[0].name)
````

````
A[3]=Player("Gill",77)
A[4]=Player("Pant",17)

for i in range(capacity):
    print(i,A[i].name if A[i] else None)
````

## Summary

| Structure | Stores | Resizable | Mixed Types |
|---|---|---|---|
| `array.array` | Primitive values | No | No |
| `list` | References | Yes | Yes |
| `ctypes.py_object` | References | No | Yes |
