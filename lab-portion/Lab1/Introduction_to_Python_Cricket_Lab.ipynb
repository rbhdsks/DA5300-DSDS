## 1. Strings

Strings are immutable sequences of characters.

```
name="Virat Kohli"
print(name)
print(len(name))
print(name.upper())
print(name.lower())
print(name[0], name[-1])
print(name[:5])
country="India"
print(name+" plays for "+country)
```

## 2. Lists

Lists are mutable collections.

```
players=["Virat","Rohit","Bumrah"]
print(players)
players.append("Gill")
players.remove("Rohit")
for p in players:
    print(p)
```

## 3. Tuples

Tuples are immutable collections.

```
formats=("TEST","ODI","T20")
print(formats)
for f in formats:
    print(f)
```

## 4. Dictionaries

Dictionaries store key-value pairs.

```
player={"Name":"Virat","Country":"India","Age":36}
print(player["Name"])
player["Role"]="Batter"
for k,v in player.items():
    print(k,v)
```

## 5. File I/O

Writing and reading text files.

```
with open("players.txt","w") as f:
    f.write("Virat\nRohit\nGill\n")
with open("players.txt") as f:
    for line in f:
        print(line.strip())
```

## 6. Classes and Objects

Creating a simple class.

```
class Player:
    def __init__(self,name,country):
        self.name=name
        self.country=country
    def display(self):
        print(self.name,self.country)
p=Player("Virat","India")
p.display()
```

## 7. Inheritance

Player inherits from Person.

```
class Person:
    def __init__(self,name):
        self.name=name
class Player(Person):
    def __init__(self,name,country):
        super().__init__(name)
        self.country=country
p=Player("Virat","India")
print(p.name,p.country)
```

## 8. Composition

Player contains a Statistics object.

```
class Statistics:
    def __init__(self,avg):
        self.avg=avg
class Player:
    def __init__(self,name,stats):
        self.name=name
        self.stats=stats
p=Player("Virat",Statistics(52.4))
print(p.name,p.stats.avg)
```

## 9. Class Methods

Class variables and class methods.

```
class Player:
    total=0
    def __init__(self,name):
        self.name=name
        Player.total+=1
    @classmethod
    def count(cls):
        print(cls.total)
Player("Virat"); Player("Rohit"); Player("Gill")
Player.count()
```

## 10. Iterators

Using iter() and next().

```
players=["Virat","Rohit","Gill"]
it=iter(players)
print(next(it))
print(next(it))
print(next(it))
for p in players:
    print(p)
```

## 11. Generators

Using yield.

```
def even_numbers():
    for i in range(10):
        if i%2==0:
            yield i
for x in even_numbers():
    print(x)

def player_gen():
    yield "Virat"
    yield "Rohit"
    yield "Gill"
for p in player_gen():
    print(p)
```
