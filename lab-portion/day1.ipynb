1. Write a function count_even(A) that returns the number of even integers in a Python list. Test it on an
empty list, one element, and a mixed list.

```python
def count_even(A):
    count=0
    for value in A:
        if value %2==0:
            count +=1
    return count
print(count_even([2,3,4,5,6,7,9]))
```

2. Reverse a list without reverse() or slicing

```python
def reverse_list(A):
    result=[]
    for index in range(len(A) -1,-1,-1):
        result.append(A[index])

    return result
print(reverse_list([2,3,4,5,6,7,8,9]))
```

```python
def second_largest(A):
    largest= None 
    second=None

    for value in A:
        if largest is None:
            largest = value
        elif value >largest:
            second = largest 
            largest=value 
        elif value < largest:
            if second is None or value > largest:
                second = value
    return second


print(second_largest([2,3,4,56,8,,9]))
```
