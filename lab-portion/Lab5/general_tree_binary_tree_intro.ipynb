# General Trees and Binary Trees -- Preparation
Simple examples for node creation, file reading, and regular expressions.

## 1. General Tree Node

```
class Node:
    def __init__(self, name):
        self.name = name
        self.children = []

root = Node("Data Structures")
chapter = Node("1. Trees")
root.children.append(chapter)

print(root.name)
print(root.children[0].name)
```

## 2. Reading a Text File

```
with open("sample_book.txt", "w") as f:
    f.write("Data Structures\n")
    f.write("2. Trees\n")
    f.write("1. Arrays\n")
    f.write("2.1 General Trees : 5\n")

with open("sample_book.txt") as f:
    for line in f:
        print(repr(line), "->", line.strip())
```

## 3. Regex: Separate Number and Title
The expression below recognizes labels such as `2.4.2.1` and also accepts an optional final period.

```
import re

pattern = r'^([0-9]+(?:\\.[0-9]+)*)\\.?\\s+(.+)$'

examples = [
    "2.4.2.1 Recursion",
    "1. Chapter Name",
    "2.4 Section Name",
    "Data Structures"
]

for line in examples:
    m = re.match(pattern, line)
    if m:
        print("number:", m.group(1), " title:", m.group(2))
    else:
        print("book title:", line)
```

## 4. Find the Level

```
def level(number):
    return len(number.split("."))

for x in ["1", "2.4", "2.4.2", "2.4.2.1"]:
    print(x, "->", level(x))
```

## 5. Find the Parent

```
def parent_number(number):
    parts = number.split(".")
    if len(parts) == 1:
        return None
    return ".".join(parts[:-1])

for x in ["1", "2.4", "2.4.2", "2.4.2.1"]:
    print(x, "-> parent:", parent_number(x))
```

## 6. Regex Including Page Count

```
line = "2.4.2.1 Recursion : 7"

pattern = r'^([0-9]+(?:\\.[0-9]+)*)\\.?\\s+(.+?)\\s*:\\s*(\\d+)\\s*$'
m = re.match(pattern, line)

print("number:", m.group(1))
print("title :", m.group(2))
print("pages :", int(m.group(3)))
```

## 7. Binary Tree Node

```
class BinaryNode:
    def __init__(self, value):
        self.value = value
        self.left = None
        self.right = None

root = BinaryNode("A")
root.left = BinaryNode("B")
root.right = BinaryNode("C")

print(root.value, root.left.value, root.right.value)
```
